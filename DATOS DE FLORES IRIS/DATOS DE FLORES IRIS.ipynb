```python
import pandas as pd
import numpy as np

# 🌱 DATOS DE FLORES IRIS (clásico, ideal para empezar)
url_iris = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
iris = pd.read_csv(url_iris)
print("=== DATOS DE FLORES IRIS ===")
print(iris.head())
print("\n📏 Medidas disponibles: largo/ancho de sépalo y pétalo")
print("🎯 Objetivo: identificar especie de flor")
```

```python
# 📈 Ver cuántas flores hay de cada especie
print("📊 Cantidad por especie:")
print(iris["species"].value_counts())

# 🧮 Calcular promedios por especie
print("\n📏 Promedios de medidas por especie:")
print(iris.groupby("species").mean(numeric_only=True))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Estilo bonito
sns.set_style("whitegrid")

# Datos promedio que ya calculamos
promedios = iris.groupby("species").mean(numeric_only=True)

# Crear gráfico
promedios.plot(kind="bar", figsize=(8, 5))
plt.title("📏 Medidas Promedio por Especie de Flor")
plt.ylabel("Tamaño (cm)")
plt.xlabel("Especie")
plt.xticks(rotation=0)
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=iris,
    x="petal_length",
    y="petal_width",
    hue="species",
    s=80  # Tamaño de los puntos
)
plt.title("🌸 Relación: Largo vs Ancho del Pétalo")
plt.xlabel("Largo del pétalo (cm)")
plt.ylabel("Ancho del pétalo (cm)")
plt.legend(title="Especie")
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.boxplot(
    data=iris,
    x="species",
    y="petal_length",
    palette="Set2"
)
plt.title("📦 Distribución del Largo del Pétalo por Especie")
plt.xlabel("Especie")
plt.ylabel("Largo del pétalo (cm)")
plt.show()
```
